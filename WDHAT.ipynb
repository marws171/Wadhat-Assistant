```python
!pip install ctransformers

from ctransformers import AutoModelForCausalLM

llm = AutoModelForCausalLM.from_pretrained(
    "Omartificial-Intelligence-Space/ALLaM-7B-Instruct-preview-Q4_K_M-GGUF",
    model_file="allam-7b-instruct-preview-q4_k_m.gguf",
    model_type="llama"
)

system_prompt = """
أنت مساعد ذكي يُدعى "وضحت". عندما يقول المستخدم "أهلاً"، رحب به بالقول: "أهلاً! أنا وضحت، مساعدك الذكي للخدمات الحكومية. كيف أساعدك؟" إذا طرح المستخدم سؤالاً، أجب بدقة ووضوح باللغة العربية الفصحى، واختم كل إجابة بـ "وضحت؟".
"""

user_greeting = "أهلاً"
greeting_prompt = f"{system_prompt}\nالمستخدم: {user_greeting}"
greeting_response = llm(greeting_prompt, max_new_tokens=100, temperature=0.7)

print(greeting_response)

user_question = "لوسمحت ممكن اعرف كيف احمل الصك الالكتروني من ناجز؟"
question_prompt = f"{system_prompt}\nالمستخدم: {user_question}"
question_response = llm(question_prompt, max_new_tokens=200, temperature=0.7)

print(question_response)
```
